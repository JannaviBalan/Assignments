```python
import numpy as np

# Create tensors
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# 1. Tensor Addition
addition = A + B
print("Tensor Addition:\n", addition)

# 2. Tensor Subtraction
subtraction = A - B
print("\nTensor Subtraction:\n", subtraction)

# 3. Element-wise Multiplication
elementwise_mul = A * B
print("\nElement-wise Multiplication:\n", elementwise_mul)

# 4. Matrix Multiplication
matrix_mul = np.matmul(A, B)
print("\nMatrix Multiplication:\n", matrix_mul)

# 5. Element-wise Division
division = A / B
print("\nElement-wise Division:\n", division)
```

```python
import numpy as np

# Create tensor
tensor = np.arange(24)
print("Original Tensor (24,):")
print(tensor)
print("Shape:", tensor.shape)

# 1. Reshape to (4,6)
tensor_4x6 = tensor.reshape(4,6)
print("\nReshaped to (4,6):")
print(tensor_4x6)
print("Shape:", tensor_4x6.shape)

# 2. Reshape to (2,3,4)
tensor_2x3x4 = tensor.reshape(2,3,4)
print("\nReshaped to (2,3,4):")
print(tensor_2x3x4)
print("Shape:", tensor_2x3x4.shape)

# 3. Convert back to (24,)
tensor_back = tensor_2x3x4.reshape(24,)
print("\nConverted back to (24,):")
print(tensor_back)
print("Shape:", tensor_back.shape)

# 4. Transpose the (4,6) tensor
transpose_tensor = tensor_4x6.T
print("\nTranspose of (4,6) tensor:")
print(transpose_tensor)
print("Shape:", transpose_tensor.shape)
```

```python
import numpy as np

# Create vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# 1. Stack them vertically
vertical_stack = np.vstack((a, b))
print("Vertical Stack:")
print(vertical_stack)

# 2. Stack them horizontally
horizontal_stack = np.hstack((a, b))
print("\nHorizontal Stack:")
print(horizontal_stack)

# 3. Combine them into a 2x3 matrix
matrix_2x3 = np.array([a, b])
print("\n2x3 Matrix:")
print(matrix_2x3)
```

```python
import numpy as np

# Create RGB image tensor
image = np.random.randint(0, 255, (224, 224, 3))

# 1. Print the tensor shape
print("Image Shape:", image.shape)

# 2. Extract the Red channel
red_channel = image[:, :, 0]
print("\nRed Channel Shape:", red_channel.shape)

# 3. Extract the Green channel
green_channel = image[:, :, 1]
print("\nGreen Channel Shape:", green_channel.shape)

# 4. Extract the Blue channel
blue_channel = image[:, :, 2]
print("\nBlue Channel Shape:", blue_channel.shape)

# 5. Convert to grayscale (average of channels)
grayscale = image.mean(axis=2)

print("\nGrayscale Image Shape:", grayscale.shape)
print("\nGrayscale Tensor:")
print(grayscale)
```

```python
import numpy as np

# Create tensor
tensor = np.arange(48)
print("Original Tensor (1D):")
print(tensor)
print("Shape:", tensor.shape)

# 1. Reshape to (6,8)
tensor_6x8 = tensor.reshape(6,8)
print("\nReshaped to (6,8):")
print(tensor_6x8)
print("Shape:", tensor_6x8.shape)

# 2. Reshape to (4,4,3)
tensor_4x4x3 = tensor.reshape(4,4,3)
print("\nReshaped to (4,4,3):")
print(tensor_4x4x3)
print("Shape:", tensor_4x4x3.shape)

# 3. Reshape to (2,3,4,2)
tensor_2x3x4x2 = tensor.reshape(2,3,4,2)
print("\nReshaped to (2,3,4,2):")
print(tensor_2x3x4x2)
print("Shape:", tensor_2x3x4x2.shape)

# 4. Flatten the tensor back to 1D
tensor_flat = tensor_2x3x4x2.reshape(-1)
print("\nFlattened back to 1D:")
print(tensor_flat)
print("Shape:", tensor_flat.shape)
```

```python
import numpy as np

# Create tensors
A = np.ones((3,3))
B = np.zeros((3,3))

# 1. Concatenate row-wise (stack B below A)
row_concat = np.vstack((A, B))
print("Row-wise Concatenation:")
print(row_concat)
print("Shape:", row_concat.shape)

# 2. Concatenate column-wise (stack B to the right of A)
col_concat = np.hstack((A, B))
print("\nColumn-wise Concatenation:")
print(col_concat)
print("Shape:", col_concat.shape)

# 3. Stack them to create a 3D tensor
stack_3d = np.stack((A, B), axis=0)  # Axis 0 adds a new dimension
print("\nStacked 3D Tensor:")
print(stack_3d)
print("Shape:", stack_3d.shape)
```

```python
import numpy as np

# Create tensor
tensor = np.arange(24).reshape(6,4)
print("Original Tensor:\n", tensor)

# 1. Split the tensor into two equal parts (along rows)
split_two = np.array_split(tensor, 2, axis=0)
print("\nSplit into two equal parts:")
for i, part in enumerate(split_two):
    print(f"Part {i+1}:\n", part)

# 2. Split the tensor into three parts (along rows)
split_three = np.array_split(tensor, 3, axis=0)
print("\nSplit into three parts:")
for i, part in enumerate(split_three):
    print(f"Part {i+1}:\n", part)

# 3. Extract the first three rows
first_three_rows = tensor[:3, :]
print("\nFirst three rows:\n", first_three_rows)

# 4. Extract the last column
last_column = tensor[:, -1]
print("\nLast column:\n", last_column)
```

```python
import numpy as np

# 1. Create the tensor
tensor = np.random.randint(1, 10, (3, 4, 2))
print("Original Tensor:\n", tensor)
print("Shape:", tensor.shape)

# 2. Transpose the tensor
transposed = tensor.transpose()  # reverses the axes by default
print("\nTransposed Tensor:\n", transposed)
print("Shape:", transposed.shape)

# 3. Swap axes 0 and 2
swap_0_2 = np.swapaxes(tensor, 0, 2)
print("\nTensor after swapping axes 0 and 2:\n", swap_0_2)
print("Shape:", swap_0_2.shape)

# 4. Swap axes 1 and 2
swap_1_2 = np.swapaxes(tensor, 1, 2)
print("\nTensor after swapping axes 1 and 2:\n", swap_1_2)
print("Shape:", swap_1_2.shape)
```

```python
import numpy as np

# Create tensor
tensor = np.random.randint(1, 100, (4,5,3))
print("Tensor:\n", tensor)
print("Shape:", tensor.shape)

# 1. Mean of the entire tensor
mean_total = tensor.mean()
print("\nMean of entire tensor:", mean_total)

# 2. Mean along axis 0 (across the 4 matrices)
mean_axis0 = tensor.mean(axis=0)
print("\nMean along axis 0:\n", mean_axis0)
print("Shape:", mean_axis0.shape)

# 3. Mean along axis 1 (across the 5 rows in each matrix)
mean_axis1 = tensor.mean(axis=1)
print("\nMean along axis 1:\n", mean_axis1)
print("Shape:", mean_axis1.shape)

# 4. Maximum values in the tensor
max_val = tensor.max()
print("\nMaximum value in tensor:", max_val)

# 5. Minimum values in the tensor
min_val = tensor.min()
print("Minimum value in tensor:", min_val)

# 6. Standard deviation
std_val = tensor.std()
print("Standard deviation:", std_val)
```

```python
import numpy as np

# Create tensor
tensor = np.random.randint(1, 50, (5,5))
print("Original Tensor:\n", tensor)

# 1. Extract all values greater than 25
greater_than_25 = tensor[tensor > 25]
print("\nValues greater than 25:\n", greater_than_25)

# 2. Replace values less than 10 with 0
tensor[tensor < 10] = 0
print("\nTensor after replacing values < 10 with 0:\n", tensor)

# 3. Count how many values are greater than 30
count_gt_30 = np.sum(tensor > 30)
print("\nNumber of values greater than 30:", count_gt_30)

# 4. Find indices where values are greater than 40
indices_gt_40 = np.argwhere(tensor > 40)
print("\nIndices where values are greater than 40:\n", indices_gt_40)
```

```python
import numpy as np

# Create tensor
tensor = np.arange(16).reshape(4,4)
print("Original Tensor:\n", tensor)

# 1. Flip the tensor horizontally (left-right)
flip_horizontal = np.fliplr(tensor)
print("\nHorizontally Flipped Tensor:\n", flip_horizontal)

# 2. Flip the tensor vertically (up-down)
flip_vertical = np.flipud(tensor)
print("\nVertically Flipped Tensor:\n", flip_vertical)

# 3. Rotate the tensor 90 degrees clockwise
rotate_90 = np.rot90(tensor, k=-1)  # k=-1 rotates clockwise
print("\nRotated 90 degrees clockwise:\n", rotate_90)

# 4. Rotate the tensor 180 degrees
rotate_180 = np.rot90(tensor, k=2)
print("\nRotated 180 degrees:\n", rotate_180)
```
