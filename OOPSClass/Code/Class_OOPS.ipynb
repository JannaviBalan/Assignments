```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def get_details(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: ₹{self.price}")
        print("-" * 30)


# Creating 3 book objects
book1 = Book("The Alchemist", "Paulo Coelho", 399)
book2 = Book("Harry Potter", "J.K. Rowling", 599)
book3 = Book("Rich Dad Poor Dad", "Robert Kiyosaki", 450)

# Displaying details
book1.get_details()
book2.get_details()
book3.get_details()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, percent):
        increase_amount = self.salary * (percent / 100)
        self.salary += increase_amount


# Create an employee
emp1 = Employee("John", 50000)

# Increase salary by 10%
emp1.increase_salary(10)

# Print updated salary
print(f"Employee Name: {emp1.name}")
print(f"Updated Salary: ₹{emp1.salary}")
```

```python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: ₹{amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance! Withdrawal denied.")
        elif amount <= 0:
            print("Invalid withdrawal amount")
        else:
            self.balance -= amount
            print(f"Withdrawn: ₹{amount}")

    def show_balance(self):
        print(f"Current Balance: ₹{self.balance}")


# Example usage
account = BankAccount(1000)

account.deposit(500)
account.withdraw(300)
account.withdraw(1500)  # should prevent negative balance

account.show_balance()
```

```python
class TemperatureMonitor:
    def check_temp(self, temp):
        if temp > 50:
            print("High")
        elif temp < 10:
            print("Low")
        else:
            print("Normal")


# Example usage
monitor = TemperatureMonitor()

monitor.check_temp(60)  # High
monitor.check_temp(5)   # Low
monitor.check_temp(25)  # Normal
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []  # list to store (name, price)

    def add_item(self, name, price):
        self.items.append((name, price))
        print(f"Added: {name} - ₹{price}")

    def total_cost(self):
        total = sum(price for name, price in self.items)
        return total


# Example usage
cart = ShoppingCart()

cart.add_item("Shirt", 800)
cart.add_item("Shoes", 1500)
cart.add_item("Watch", 2000)

# Calculate total
print(f"Total Cost: ₹{cart.total_cost()}")
```

```python
class Team:
    def __init__(self):
        self.players = []  # list to store player names

    def add_player(self, name):
        self.players.append(name)
        print(f"{name} added to the team")

    def show_team(self):
        print("Team Players:")
        for player in self.players:
            print(player)


# Example usage
team = Team()

team.add_player("Virat")
team.add_player("Rohit")
team.add_player("Dhoni")

team.show_team()
```

```python
class Validator:
    def is_email_valid(self, email):
        if "@" in email:
            return True
        else:
            return False


# Example usage
validator = Validator()

emails = ["user@example.com", "invalidemail.com", "hello@domain"]

for e in emails:
    if validator.is_email_valid(e):
        print(f"{e} → Valid")
    else:
        print(f"{e} → Invalid")
```

```python
class Formatter:
    def format_name(self, name):
        return name.upper()


# Example usage
formatter = Formatter()

names = ["Rani Lakshmi Bai", "Aruna Asaf Ali", "Mahatma Gandhi"]

for n in names:
    formatted_name = formatter.format_name(n)
    print(formatted_name)
```

```python
class Order:
    def __init__(self, order_id, amount):
        self.order_id = order_id
        self.amount = amount

    def apply_discount(self, percent):
        discount_amount = self.amount * (percent / 100)
        self.amount -= discount_amount
        print(f"Discount of {percent}% applied. New amount: ₹{self.amount}")


# Example usage
order1 = Order("ORD101", 2000)

order1.apply_discount(10)  # Apply 10% discount
```

```python
class DeliveryService:
    def estimate_time(self, distance):
        if distance < 5:
            return "30 mins"
        elif 5 <= distance <= 20:
            return "1 hour"
        else:
            return "2 hours"


# Example usage
service = DeliveryService()

distances = [3, 10, 25]

for d in distances:
    print(f"Distance: {d} km → Estimated Time: {service.estimate_time(d)}")
```
