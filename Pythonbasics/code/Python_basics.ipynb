```python
def is_prime(n):
    """
    Returns True if n is a prime number, False otherwise.
    """
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    # Check divisibility from 3 up to sqrt(n)
    for i in range(3, int(n**0.5)+1, 2):
        if n % i == 0:
            return False
    return True

# Example usage
num = 29
if is_prime(num):
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
def factorial(n):
    """
    Returns the factorial of n (n!).
    """
    if n < 0:
        return "Factorial is not defined for negative numbers."
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

# Example usage
num = 5
print(f"The factorial of {num} is {factorial(num)}")
```

```python
def reverse_string(s):
    """
    Returns the reverse of the input string s.
    """
    return s[::-1]  # Slice notation to reverse

# Example usage
text = "Hello, World!"
reversed_text = reverse_string(text)
print("Original String:", text)
print("Reversed String:", reversed_text)
```

```python
# Initialize empty contact book
contact_book = {}

def add_contact(name, phone):
    contact_book[name] = phone
    print(f"Contact added: {name} -> {phone}")

def search_contact(name):
    if name in contact_book:
        print(f"{name}'s number: {contact_book[name]}")
    else:
        print(f"{name} not found in contact book.")

def delete_contact(name):
    if name in contact_book:
        del contact_book[name]
        print(f"{name} has been deleted from the contact book.")
    else:
        print(f"{name} not found in contact book.")

# Example usage
while True:
    print("\nContact Book Menu:")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Delete Contact")
    print("4. Show All Contacts")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == '1':
        name = input("Enter contact name: ")
        phone = input("Enter phone number: ")
        add_contact(name, phone)

    elif choice == '2':
        name = input("Enter contact name to search: ")
        search_contact(name)

    elif choice == '3':
        name = input("Enter contact name to delete: ")
        delete_contact(name)

    elif choice == '4':
        print("\nAll Contacts:")
        for name, phone in contact_book.items():
            print(f"{name}: {phone}")

    elif choice == '5':
        print("Exiting Contact Book. Goodbye!")
        break

    else:
        print("Invalid choice! Please select 1-5.")
```

```python
# Ask user for file name
file_name = input("Enter the file name: ")

# Open the file
with open(file_name, "r") as file:
    lines = file.readlines()

# Count number of lines
line_count = len(lines)

# Count number of words
word_count = 0
for line in lines:
    words = line.split()
    word_count += len(words)

# Print results
print("Number of lines:", line_count)
print("Number of words:", word_count)
```
