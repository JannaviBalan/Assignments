```python
import numpy as np
import matplotlib.pyplot as plt

# Data
X = [1, 2, 3, 4, 5]
Y = [2, 4, 6, 8, 10]

# Create line plot with markers
plt.plot(X, Y, marker='o')

# Label axes
plt.xlabel("X values")
plt.ylabel("Y values")

# Add title
plt.title("Line Plot of X vs Y")

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
X = [1, 2, 3, 4, 5]
Y1 = [1, 4, 9, 16, 25]
Y2 = [1, 8, 27, 64, 125]

# Plot lines with different styles
plt.plot(X, Y1, linestyle='-', marker='o', label='Y1 (Squares)')
plt.plot(X, Y2, linestyle='--', marker='s', label='Y2 (Cubes)')

# Labels and title
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Comparison of Two Lines")

# Add legend
plt.legend()

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
categories = ["Food", "Rent", "Travel", "Savings"]
values = [30, 40, 10, 20]

# Explode one slice (e.g., Rent)
explode = [0, 0.1, 0, 0]

# Create pie chart
plt.pie(values, labels=categories, autopct='%1.1f%%', explode=explode)

# Add title
plt.title("Monthly Expenses Distribution")

# Show chart
plt.show()
```

```python
1.	When should you use:
o	Line plot vs Bar chart?
        Use a line plot when you want to show:
LINE PLOT:
Trends over time or sequence
Continuous data
How values change smoothly

Example:

Temperature over days
Study hours vs scores over time
Stock prices

BAR PLOT:

Different categories
Discrete groups
Separate values

Example:

Marks in different subjects
Sales of different products
Population of countries
2.	What does a histogram tell you?

A histogram shows the distribution of data.

It helps you understand:

How data is spread
Where most values lie
Frequency of values in ranges (bins)
Example insights:
Are values concentrated in the middle?
Is data skewed (left or right)?
Is it evenly distributed or clustered?

3.	Why is scatter plot useful in Machine Learning?

A scatter plot shows relationship between two variables.

In Machine Learning, it helps to:

Identify correlations (positive/negative/no relation)
Detect patterns in data
Spot outliers (unusual points)
Understand feature relationships before training models
Example:
Study hours vs exam scores
House size vs price

4.	What happens if you don’t label axes properly?

If axes are not labeled:

 Graph becomes confusing
 Viewers may misunderstand data
 Hard to interpret meaning
 Reduces presentation quality


```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
students = ["Aravind", "Brindha", "Charumathy", "Dilip", "Eswar"]

math = [85, 78, 92, 70, 88]
science = [90, 80, 85, 75, 95]
english = [88, 76, 89, 80, 84]

colors=["pink","green","purple"]

x = np.arange(len(students))  # positions for bars
width = 0.25  # bar width

# Create figure
plt.figure(figsize=(10, 5))

# Bar chart (comparison)
plt.bar(x - width, math, width, label="Math", color="pink")
plt.bar(x, science, width, label="Science", color="yellow")
plt.bar(x + width, english, width, label="English" , color="purple")

# Line plot (trend)
plt.plot(x, math, marker='o', linestyle='-', label="Math Trend")
plt.plot(x, science, marker='s', linestyle='--', label="Science Trend")
plt.plot(x, english, marker='^', linestyle='-.', label="English Trend")

# Labels and title
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Performance Dashboard")

# X-axis labels
plt.xticks(x, students)

# Legend
plt.legend()

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer

# Load dataset
data = load_breast_cancer()

df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

# target: 0 = malignant, 1 = benign

# Graph 1: Count of Benign vs Malignant (Bar Chart)

df["target"].value_counts().plot(kind="bar", color=["red", "green"])

plt.title("Cancer Type Distribution")
plt.xlabel("Type (0 = Malignant, 1 = Benign)")
plt.ylabel("Count")

plt.show()

# Graph 2 : Mean Radius Distribution (Histogram)
plt.hist(df["mean radius"], bins=20, color="skyblue", edgecolor="black")

plt.title("Distribution of Mean Radius")
plt.xlabel("Mean Radius")
plt.ylabel("Frequency")

plt.show()

# Graph 3: Relationship (Scatter Plot)

plt.scatter(df["mean radius"], df["mean texture"],
            c=df["target"], cmap="coolwarm")

plt.title("Mean Radius vs Mean Texture")
plt.xlabel("Mean Radius")
plt.ylabel("Mean Texture")

plt.show()
```
